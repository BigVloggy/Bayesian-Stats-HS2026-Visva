```python
import numpy as np

# domain (length)
a = 0
b = np.pi

n = 1000000 # number of uniformely distributed random points


x_values = np.random.uniform(a, b, size=(n, 1))
y_values = np.sin(x_values) #put in here function! For example f(x) = sin(x)



k = np.sum(y_values)

average_height = k/n
length = b - a

integral = average_height * length
integral
```

```python
###### numpy Tutorial
import numpy as np

# Python list: [1, 2, 3, 4]
# NumPy array: [1 2 3 4] --> Das ist eine Art Vektor! Nicht mehr nur eine reine Liste daher gut mathematisch manipulierbar!

numpy_array = np.array([1, 2, 3, 4, 5,]) #vectorized array
print("Numpy array (vectorized):", numpy_array)

print("Numpy array (vectorized) and added:", numpy_array + 10)

print("Done")


a = np.array ([
    (1.5, 2 , 3), 
    (4, 5, 6),
    (0, 0, 0),
    (1, 1, 1)
    ])

#a ist eine 4x3 Matrix. Schreib immer pro Zeile (a_11, a_12, a_13), und dann darunter (oder daneben geht auch) die nächste Zeile
print(a)

print("Done")
arr2 = np.arange(0,10,2) # start, stop, step
print(arr2)

print("Done")


arr3 = np.linspace(0, 1, 5) # start, stop, number of elements (evenly spaced numbers
print(arr3)

zeros = np.zeros(5)
print(zeros)

ones = np.ones(5)
print(ones)

full = np.full(5, 7) #shape (number of elements), fill_value
print(full)


### Identity Matrix
identity = np.eye(4)
print(identity)


print("Done")

############# Random Arrays
random_arr = np.random.rand(5) # Uniform distribution [0,1) with length of array = 5
print(random_arr)


random_int = np.random.randint(1, 10, size = 5) # [low, high), size
print(random_int)

print("Done")


zeros_2d = np.zeros((3, 5)) #3 rows, 5 columns
print(zeros_2d)

print("Done")

ones_3d = np.ones((2, 3, 4)) # 2x3x4 array
print(ones_3d)

print("Done now k")

k = np.array([
    (1, 0, 0),
    (1, 0, 0),
    (1, 0, 0)
])
print(k)

print("Done now k^T")


l = np.transpose(k)
print(l)


print("Done")
j = np.array([1, 2, 3])
print(np.mean(j))
print(np.sin(j))


m = np.sqrt(2)
print("Done square root of two =", m)
```
